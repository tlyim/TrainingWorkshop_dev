### Conditional Statements in Python

````python
# Basic conditional statement using if ... else ...

if Year == 2023:
    print("The value of the Year variable is 2023!")
else:
    print(f"The value of the Year variable is {Year}, not 2023!")

# Note: The f before the string in the print() function allows embedding variable values directly within the string using curly braces {}
````

````python
# If statement without else

if Year == 2023:
    print("The value of the Year variable is 2023!") 

Year   # Display the value of the variable Year again
````

````python
# If statement without else is the same as including an else part that does nothing

if Year == 2023:
    print("The value of the Year variable is 2023!")
else:
    None   # Do nothing in the else part

# Note: None is a special constant in Python that represents the absence of a value or a null value
#       None is different from the character string 'None'
````

### Control flow statements in Python

- For-loop lets you automate repetitive tasks!

````python
for yr in years:
    # You can do more complicated operations than this inside the loop
    print(yr)
````

- Pairing for-loops with conditional statements allows conditional automation!

````python
for yr in years:
    if yr == 2023:
        print("Found 2023!")
    print(yr)

# The print(yr) line is outside the if block, so it runs for every year in the list
````

- Indentation is important in Python!

````python
for yr in years:
    if yr == 2023:
        print("Found 2023!")
        print(yr)

# The print(yr) line is inside the if block, so it only runs when year is 2023
````

## A Quick Overview of Variables and Data Types in Python


In Python, variables are containers used to store data values, while data types define the kind of data a variable holds.

### Python Variables

A variable is created the moment you first assign a value to it using the assignment operator (`=`).

**Dynamic Typing:** You do not need to declare a variable's type. Python automatically infers it from the value assigned.

**Reassignment:** Variables can change types during execution. For example, a variable holding an integer can later be reassigned a string.

#### Naming Rules

- Must start with a letter or an underscore (`_`)
- Cannot start with a number
- Can only contain alphanumeric characters and underscores
- Are case-sensitive (e.g., `age` and `Age` are different)
- Cannot use Python reserved keywords like `if`, `else`, or `class`

### Core Data Types

Python's built-in data types are categorized into several groups:

#### Numeric Types

- **int:** Whole numbers (e.g., `10`, `-5`)
- **float:** Decimal numbers (e.g., `3.14`)
- **complex:** Numbers with real and imaginary parts (e.g., `3 + 5j`)

#### Sequence Types

- **str:** Text data enclosed in quotes
- **list:** Ordered, mutable collections (e.g., `[1, "apple", 3.5]`)
- **tuple:** Ordered, immutable collections (e.g., `(10, 20)`)

#### Mapping Type

- **dict:** Unordered collections of key-value pairs (e.g., `{"name": "Alice"}`)

#### Set Types

- **set:** Unordered collections of unique elements

#### Boolean Type

- **bool:** Represents `True` or `False`

#### None Type

- **NoneType:** Represents the absence of a value (`None`)

### Type Conversion & Checking

**Check Type:** Use the `type()` function (e.g., `type(x)`) to find a variable's data type.

**Type Casting:** Convert between types using functions like `int()`, `float()`, or `str()`.

**Type Hinting:** For better readability in 2026, developers often use Type Hints (e.g., `age: int = 25`) to indicate intended data types, which can be checked by tools like Mypy.

#### Undefined values from computation will throw an error, **disrupting** the execution of a code!

````python
market_cap = 80000
total_liab = 0

# compute market capitalization to total liabilities ratio 
market_cap / total_liab
````

#### _**Solution**_: Exception handling with `try` and `except`

````python


print("market_cap =", market_cap, "; total_liab =", total_liab)
print("market_cap to total_liab ratio (MV2TL) = ", market_cap, "/", total_liab, "\n")

try:
    MV2TL = market_cap / total_liab
except ZeroDivisionError:
    MV2TL = float('nan')  # Assign NaN if division by zero occurs

print("MV2TL =", MV2TL)  # Display the value of MV2TL
````

_**Note:**_

- Sometimes, data are not available for all companies in a year (e.g., due to **delisting** following mergers and acquisitions)
- Errors resulting from **unavailable data** can also be handled with `try` and `except` similarly

### The Primitive Form (Compute Z-Score with Hardcoded Data)

- Define **a function** to compute Altman Z-Score
- Compute Z-Score with Hardcoded Data

**Example:**

#### Define a function `zscore()` to calculate Altman Z-Score from input data

_**Note:**_

In practice, the computation of Altman's Z-Score uses only the **market value of common shares** (aka. **market capitalization**), instead of both preferred and common shares, because

- Preferred stock is rare among public companies;
- When issued, preferred shares are often not publicly traded (or have negligible liquidity), making a reliable market price unavailable.

````python
def zscore(
    total_assets,
    current_assets,
    current_liab,
    retained_earnings,
    ebit,
    total_liab,
    sales,
    market_cap,
    ):  

    # Altman Z-Score Components
    WkCap2TA = (current_assets - current_liab) / total_assets
    RE2TA = retained_earnings / total_assets
    EBIT2TA = ebit / total_assets
    preferred_market_cap = 0  # assuming no preferred stock for simplification
    # Essentially, use only market capitalization (i.e., MV of Common Stock) for MVofStock below:
    MVofStock2BVofTL = (market_cap + preferred_market_cap) / total_liab 
    Sales2TA = sales / total_assets

    # Formula
    zscore = 1.2*WkCap2TA + 1.4*RE2TA + 3.3*EBIT2TA + 0.6*MVofStock2BVofTL + 1.0*Sales2TA

    return zscore
````

### Compute Altman Z-Score with Hypothetical Company Data (Hardcoded)

````python

zscore_0 = zscore(
    total_assets=100000,
    current_assets=40000,
    current_liab=20000,
    retained_earnings=30000,
    ebit=15000,
    total_liab=50000,
    sales=120000,
    market_cap=80000,
    )

print(f"The Z-Score is: {zscore_0:.2f}")
````
